````python
def load_csvs():
    data = {}   # important!
    for key, path in csv.items():

        # detect delimiter based on file name
        if "customer" in path:
            sep = ';'
       ## elif "marketing" in path:   # <-- your marketing_campaign.tsv
            sep = '\t'
        else:
            sep = ','

        print(f"Loading {path} using sep='{sep}' ...")
        df = pd.read_csv(path, sep=sep)

        print(f"  → Loaded {len(df):,} rows, {len(df.columns)} columns")
        data[key] = df

    return data
````

## **TRANSACTION_MASTER_FULL – ETL Data Dictionary**


**Key assumptions:**

- The simulated Fictitious bank (Bank ABC) is based in Estonia.

- Customers have various citizenships (as reflected in the customer master), but their accounts are based in Estonia.

- For AML simulation, about 20% of incoming transfers originate from watchlist/high-risk jurisdictions.

| Column                      | Type     | Source                            | Description                                                                                                                                                                        |
| --------------------------- | -------- | --------------------------------- | ---------------------------------------------------------------------------------------------------------------------------------------------------------------------------------- |
| `transaction_id`            | string   | Bank transactions (baseline)      | Unique identifier for each transaction. Generated or inherited from the baseline dataset.                                                                                          |
| `customer_id`               | string   | CUSTOMER_MASTER_FULL              | Identifier of the internal customer associated with the transaction. Foreign key to `CUSTOMER_MASTER_FULL`.                                                                        |
| `account_id`                | string   | Derived                           | Internal bank account number used for the transaction. Generated using an IBAN-like format (e.g., `EE88…`) and linked to `customer_id`.                                            |
| `transaction_amount`        | float    | Bank transactions                 | Monetary value of the transaction. Positive numeric value.                                                                                                                         |
| `transaction_date`          | datetime | Bank transactions                 | Timestamp when the transaction occurred. Standardized to proper datetime format.                                                                                                   |
| `transaction_type`          | string   | Bank transactions                 | High-level transaction type such as `PAYMENT`, `TRANSFER`, `CASH_WITHDRAWAL`, `CARD_PURCHASE`, etc.                                                                                |
| `location`                  | string   | Bank transactions / synthetic     | Physical or logical location of the transaction (e.g., branch, city, or “ONLINE”).                                                                                                 |
| `device_id`                 | string   | Bank transactions / synthetic     | Identifier of the device/terminal used (ATM ID, POS terminal ID, browser/device fingerprint).                                                                                      |
| `ip_address`                | string   | Bank transactions / synthetic     | IP address for online or mobile transactions. Null for offline channels (ATM/branch).                                                                                              |
| `merchant_id`               | string   | Bank transactions / synthetic     | Identifier of the merchant/payee for POS or online transactions.                                                                                                                   |
| `channel`                   | string   | Standardized from baseline        | Delivery channel normalized to values such as `ATM`, `WEB`, `MOBILE`, `BRANCH`, `POS`.                                                                                             |
| `transaction_duration`      | float    | Bank / synthetic                  | Duration of the transaction/session in seconds (if available). Optional behavioral signal.                                                                                         |
| `login_attempts`            | int      | Bank / synthetic                  | Number of login attempts before the transaction. Set to 0 for offline channels.                                                                                                    |
| `previous_transaction_date` | datetime | Derived per `customer_id`         | Timestamp of the customer's previous transaction. Adjusted so it is **never earlier** than `dt_customer` in the customer master.                                                   |
| `nameOrig`                  | string   | PaySim-inspired / synthetic       | **Originator (sender)** full name. ~80% generated via Faker to represent external counterparties; ~20% randomly sampled from customer names to simulate domestic P2P transactions. |
| `nameDest`                  | string   | PaySim-inspired / customer master | **Destination (receiver)** full name, taken from the customer's `first_name + surname`. Assumes internal customer is the beneficiary.                                              |
| `payment_currency`          | string   | SAML-D-inspired / synthetic       | Currency used to send the payment (e.g., `EUR`, `USD`, `GBP`). Majority in EUR, minority in FX.                                                                                    |
| `received_currency`         | string   | SAML-D-inspired / synthetic       | Currency received by the beneficiary. Often equal to `payment_currency`, but may differ for FX transactions.                                                                       |
| `sender_bank_location`      | string   | SAML-D-inspired / synthetic       | Country of the sender’s bank. Modeled as **80% EU/low-risk countries** and **20% high-risk/watchlist countries** to support AML inflow simulation.                                 |
| `receiver_bank_location`    | string   | SAML-D-inspired / synthetic       | Country of the receiver's bank. For this PoC, default is `Estonia` as the home country of fictional Bank ABC.                                                                      |
| `payment_type`              | string   | SAML-D-inspired / synthetic       | Payment type such as `CREDIT_TRANSFER`, `DEBIT_TRANSFER`, `CARD`, `INSTANT`, `CROSS_BORDER`.                                                                                       |


## **ETL PIPELINE OVERVIEW**

The ETL pipeline consists of eight major steps:

1. Extraction
    Load baseline transaction data (bank_transaction_data_2.csv) & customer data (CUSTOMER_MASTER_FULL.csv).
2. Standardization & Normalization
    - Clean column names
    - Parse dates (transaction_date).
    - Normalize channels → ATM, WEB, MOBILE, BRANCH, POS.
    - Cast numeric values to appropriate types.
3. Customer Linking & Account Mapping
    - Assign customer_id to each transaction.
    - Generate IBAN-like account_id and associate with customers.
    - Join dt_customer from the customer master.
    - Ensure transaction_date >= dt_customer.
4. Previous Transaction Derivation
    - Sort by customer_id and transaction_date.
    - Compute previous_transaction_date using lag.
    - Correct early values to match dt_customer.
5. Name Enrichment (PaySim-inspired)
    - nameDest: internal customer → first_name + surname.
    - nameOrig: mix of synthetic external names and 20% internal names.
6. Payment & AML Attribute Enrichment (SAML-D-inspired)
    - Generate payment_currency / received_currency.
    - Assign sender_bank_location using EU vs high-risk proportions (80/20).
    - Set receiver_bank_location = Estonia.
    - Assign payment_type based on distribution of real banking patterns.
7. Schema Alignment & Cleanup
    - Ensure all columns match the final schema.
    - Drop intermediate fields.
    - Reorder columns.
8. Export Data
    - Save dataset as TRANSACTION_MASTER_FULL.csv.
    - export to SQL Server (if needed).

### **Asumption**

| previous_transaction_date | dt_customer | max            | Explanation                                     |
| ------------------------- | ----------- | -------------- | ----------------------------------------------- |
| 2020-01-01                | 2022-05-10  | **2022-05-10** |invalid, so if previous_transaction_date < dt_customer pick the max date |
| NaT                       | 2022-05-10  | **2022-05-10** | invalid. if Null input with dt_customer |
| 2023-03-01                | 2022-05-10  | **2023-03-01** | valid, previous_transaction_date > dt_customer pick the max date|


````python
# Previous Transaction Date

#1. Short transaction based on date (oldest to newest date)
df_transactions_final = df_transactions_final.sort_values(["customer_id", "transaction_date"])

#2. take the previous transaction date for each customer and shift the data to the previous row
df_transactions_final["previous_transaction_date"] = df_transactions_final.groupby("customer_id")["transaction_date"].shift()

#3. Convert both date to datetime so that max() can run
df_transactions_final["previous_transaction_date"] = pd.to_datetime(df_transactions_final["previous_transaction_date"], errors="coerce")
df_transactions_final["dt_customer"] = pd.to_datetime(df_transactions_final["dt_customer"], errors="coerce")

# compare previous transaction date to dt_customer, and pick the max date 
df_transactions_final["previous_transaction_date"] = df_transactions_final[["previous_transaction_date", "dt_customer"]].max(axis=1)
````

````python
# mapping remapping data location

# 1. Create region & cities dictionary

cities = {
    "Estonia": [
        "Tallinn", "Tartu", "Narva", "Pärnu", "Viljandi",
        "Kuressaare", "Rakvere", "Sillamäe"
    ],
    "Europe": [
        "Helsinki", "Stockholm", "Oslo", "Copenhagen", "Berlin",
        "Paris", "Madrid", "Rome", "Warsaw", "Prague"
    ],
    "Asia": [
        "Jakarta", "Singapore", "Tokyo", "Seoul", "Bangkok",
        "Kuala Lumpur", "Dubai", "Doha", "Sharjah", "Jeddah", "Dammam", "Yangon"
    ],
    "America": [
        "New York", "Los Angeles", "San Francisco", "Chicago",
        "Toronto", "Vancouver", "Mexico City", "San Diego", "Houston"
    ],
    "Oceania": [
        "Sydney", "Melbourne" 
    ],
    "Africa": [
        "Cape Town", "Johannesburg", "Lagos","Nairobi"
    ]
}

# 2. Assign Region With Priority for Estonia

region_weights = {
    "Estonia": 0.70,
    "Europe": 0.20,
    "Asia": 0.05,
    "America": 0.05
}

# 3. Assign Region based on weight

def pick_region():
    return random.choices(
        population=list(region_weights.keys()),
        weights=list(region_weights.values()),
        k=1
    )[0]


# 4. Assign a City Based on the Region

def pick_city(region):
    return random.choice(cities[region])

# Return a Random City With Estonia Priority

def random_city():
    region = pick_region()
    return pick_city(region)

df.loc[:, "location"] = df.apply(
    lambda x: random_city(), axis=1
)
````

````python
#standardixe channel values

def clean_channel(x):
    if pd.isna(x):
        return "UNKNOWN"

    x = str(x).strip().upper()

    if "ATM" in x:
        return "ATM"
    if any(k in x for k in ["WEB", "ONLINE", "INTERNET"]):
        return "INTERNET BANKING"
    if any(k in x for k in ["MOBILE", "APP"]):
        return "MOBILE BANKING"
    if any(k in x for k in ["POS", "EDC", "CARD"]):
        return "EDC"
    if "BRANCH" in x:
        return "BRANCH"

    return x


# apply dataset df_trx

df["channel"] = df["channel"].apply(clean_channel)

df.head()
````

````python
#generate IBAN Estonia

def random_iban_estonia():
    return "EE" + "".join([str(np.random.randint(0,10)) for _ in range(18)])

# generate IBAN International

def random_iban_international():
    region = random.choice(list(country_to_country_code.keys()))
    code = random.choice(country_to_country_code[region])
    return code + "".join([str(np.random.randint(0,10)) for _ in range(20)]) #just assumption
````

````python
aml_blacklist_codes = ["IR", "SY", "KP", "SD", "MM", "RU"]

mix_choices = np.random.choice(
    ["INTERNAL", "DOMESTIC", "INTERNATIONAL"],
    size=len(df_mix),
    p=[0.65, 0.25, 0.10]  # you may adjust
)

sender_list = []

internal_accounts = df_mix["account_id"].unique().tolist()

for i, choice in enumerate(mix_choices):

    receiver_acc = df_mix.loc[i, "account_id"]

    # INTERNAL (to another ABC customer)
    if choice == "INTERNAL":
        valid_internal = [a for a in internal_accounts if a != receiver_acc]
        sender_list.append(
            random.choice(valid_internal) if valid_internal else random_iban_estonia()
        )

    # DOMESTIC (Estonia - external bank)
    elif choice == "DOMESTIC":
        sender_list.append(random_iban_estonia())

    # INTERNATIONAL (safe + AML blacklist)
    else:

        # 0.05% AML blacklist routing
        if np.random.rand() < 0.0005:
            code = random.choice(aml_blacklist_codes)
            iban = code + "".join([str(np.random.randint(0,10)) for _ in range(20)])
            sender_list.append(iban)
        else:
            sender_list.append(random_iban_international())

# assign
df_mix["sender_account"] = sender_list
````
